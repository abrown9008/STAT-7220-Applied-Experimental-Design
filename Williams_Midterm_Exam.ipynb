**PART 3 INSTRUCTIONS:** Please briefly respond to the following questions:

**Question 1.** Give a reason why one-factor CRD wouldn't generally be appropriate for studies involving human participants.

A one-factor completely randomized design is often not ideal for human studies because people differ in many important ways, such as age, experience, motivation, health, and background. These differences can affect the outcome and create extra variability if they are not controlled.



**Question 2.** What is the purpose of post-hoc tests?

Post-hoc tests are used after a significant ANOVA result to determine which specific group means differ from each other while controlling for multiple comparisons.


**Question 3.** Explain the difference between RBD and LSD.

A randomized block design (RBD) controls for one known source of variation by grouping similar experimental units into blocks before random assignment of treatments.

A Latin square design (LSD) controls for two sources of nuisance variation at the same time by arranging treatments in rows and columns so each treatment appears once in each row and once in each column.


**Question 4.** Explain why randomization is important in designing experiments.

Randomization is important because it helps reduce bias, balances lurking variables across treatment groups, and makes cause-and-effect conclusions more valid.


**Question 5.** Explain why replication is important in designing experiments.

Replication is important because it allows researchers to estimate natural variability, improves precision, increases reliability, and makes the conclusions more trustworthy.



```python
import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as stats
import statsmodels.api as sm
from statsmodels.formula.api import ols
from statsmodels.stats.multicomp import pairwise_tukeyhsd
```

```python
vg = pd.read_excel("Video Game Controllers.xlsx")
vr = pd.read_excel("Nursing VR Study.xlsx")
```

```python
# ---------------- PART 1 ----------------
print("PART 1 SUMMARY")
print(vg.groupby("Material")["Comfort"].agg(["count","mean","std","median","min","max"]))

vg.boxplot(column="Comfort", by="Material")
plt.title("Comfort by Material")
plt.suptitle("")
plt.show()

model1 = ols("Comfort ~ C(Material)", data=vg).fit()
anova1 = sm.stats.anova_lm(model1, typ=2)
print("\nPART 1 ANOVA")
print(anova1)

print("\nPART 1 SHAPIRO")
print(stats.shapiro(model1.resid))

groups1 = [g["Comfort"].values for _, g in vg.groupby("Material")]
print("\nPART 1 LEVENE")
print(stats.levene(*groups1, center="median"))

if anova1.loc["C(Material)", "PR(>F)"] < 0.05:
    print("\nPART 1 TUKEY")
    print(pairwise_tukeyhsd(vg["Comfort"], vg["Material"]))
```

```python
# ---------------- PART 2 ----------------
print("\nPART 2 SUMMARY BY METHOD")
print(vr.groupby("Method")["Score"].agg(["count","mean","std","median","min","max"]))

print("\nPART 2 SUMMARY BY EXPERIENCE")
print(vr.groupby("Experience")["Score"].agg(["count","mean","std","median","min","max"]))

print("\nPART 2 SUMMARY BY METHOD AND EXPERIENCE")
print(vr.groupby(["Method","Experience"])["Score"].agg(["count","mean","std","median","min","max"]))

vr.boxplot(column="Score", by="Method")
plt.title("Score by Method")
plt.suptitle("")
plt.show()

model2 = ols("Score ~ C(Method) + C(Experience)", data=vr).fit()
anova2 = sm.stats.anova_lm(model2, typ=2)
print("\nPART 2 ANOVA")
print(anova2)

print("\nPART 2 SHAPIRO")
print(stats.shapiro(model2.resid))

groups2 = [g["Score"].astype(float).values for _, g in vr.groupby(["Method","Experience"])]
print("\nPART 2 LEVENE")
print(stats.levene(*groups2, center="median"))

if anova2.loc["C(Method)", "PR(>F)"] < 0.05:
    print("\nPART 2 TUKEY")
    print(pairwise_tukeyhsd(vr["Score"], vr["Method"]))
```
